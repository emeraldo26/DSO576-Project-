# Phase 1 data exploration

This notebook runs `explore_data.py`.

```python
from pathlib import Path
import sys

project_root = Path.cwd()
if not (project_root / 'explore_analysis').exists():
    project_root = Path.cwd().parent
sys.path.insert(0, str(project_root / 'explore_analysis'))

from explore_data import load_data, explore
data = load_data(project_root / 'data')
occupation_table, top_tasks, report = explore(data)

output_dir = project_root / 'outputs'
output_dir.mkdir(parents=True, exist_ok=True)
occupation_table.to_csv(output_dir / 'occupation_exploration.csv', index=False)
top_tasks.to_csv(output_dir / 'top_tasks_by_occupation.csv', index=False)
(output_dir / 'exploration_report.txt').write_text(report, encoding='utf-8')
print(report)
```
